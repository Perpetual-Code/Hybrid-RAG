```python
from collections import defaultdict
```

```python
# A list of student names and their classes
student_data = [
    ("Class A", "Alice"),
    ("Class B", "Bob"),
    ("Class A", "Charlie"),
    ("Class B", "David"),
]
```

```python
student_by_class=defaultdict(list)
```

```python
for class_name,student in student_data:
    student_by_class[class_name].append(student)

print(dict(student_by_class))
```

```python
inventory = ["apple", "banana", "apple", "orange", "banana", "apple"]

inventory_data=defaultdict(int)

for item in inventory:
    inventory_data[item]+=1

print(dict(inventory_data))
```

```python
words = ["cat", "dog", "cat", "bird", "dog", "cat"]

word_count:dict[str,float]=defaultdict(float)

for word in words:
    word_count[word]+=1
print(dict(word_count))
```

```python
pairs = [
    ("fruit", "apple"),
    ("vegetable", "carrot"),
    ("fruit", "banana"),
    ("vegetable", "carrot"),
    ("vegetable", "carrot"),
]

```

```python
pairs_cat:dict[str,set[str]]=defaultdict(set)
for cat,item in pairs:
    pairs_cat[cat].add(item)

print(dict(pairs_cat))
```

```python
pair_cat:dict[str,list[str]]={}

for cat,item in pairs:
    if cat not in pair_cat:
        pair_cat[cat]=[]
        

    pair_cat[cat].append(item)


pair_cat
```
