```python
ROOT_DIR=Path.cwd()
DATA_DIR=ROOT_DIR/"data"/"fiqa"
```

```python
corpus=pd.read_parquet(DATA_DIR/"corpus.parquet")
queries=pd.read_parquet(DATA_DIR/"queries.parquet")
qrels=pd.read_parquet(DATA_DIR/"qrels.parquet")
```

```python
# Filter the queries to only include the test queries
test_queries_ids=set(qrels['query-id'].astype(str))
test_queries=queries[queries['_id'].astype(str).isin(test_queries_ids)]
print(len(test_queries))
```

```python
query=test_queries.iloc[0]
print("\nExample Query\n")
print("_id",query['_id'])
print("text",query['text'])
```

```python
relevant=qrels[qrels["query-id"].astype(str)==str(query["_id"])]
print(list(relevant["corpus-id"]))
```
