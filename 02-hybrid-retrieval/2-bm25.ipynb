```python
# Step 4 Run a query
def search_bm25(query:str,k:int=10)->list[tuple[str,float]]:
    """ Return the top k (doc_id scorr=e) pairs for a query"""
    retriever=bm25s.BM25()
    query_tokens=bm25s.tokenize([query],stopwords="en")
    result,scores=retriever.retrieve(query_tokens=query_tokens,k=k,return_as='document')

    return list(zip(result[0],scores[0].to_list()))
```
