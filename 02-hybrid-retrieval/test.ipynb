```python
import os
from dotenv import load_dotenv
load_dotenv()
from openai import OpenAI
from openrouter import OpenRouter
from langchain_nvidia_ai_endpoints import NVIDIAEmbeddings
```

```python
client=OpenAI(base_url="https://openrouter.ai/api/v1",api_key=os.getenv("OPENROUTER_API_KEY"))
```

```python
client=NVIDIAEmbeddings(model="nvidia/nemotron-3-embed-1b",api_key=os.getenv("NVIDIA_API_KEY"))
```

```python
text_to_embed=["If you prefer not to download models locally, you can access embedding models through Hugging Face Inference Providers or a dedicated Inference Endpoint. Both expose open-source embedding models on Hugging Face’s scalable serverless infrastructure."]
```

```python
result=client.embed_documents(texts=text_to_embed)
```
